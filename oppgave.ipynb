## Lage en liste

For å lage en liste lager vi en variabel og tildeler den listen. En liste i Python kan inneholde både tall og tekst, og hvert element i listen skilles med et kommategn. Starten og slutten på listen rammes inn med firkantparenteser.
En liste kan altså se slik ut: `[1, 3, 8, "hei", "hallo", 2901]`

```python
handleliste = ["Tomater", "Agurk", "Appelsin", "Sjokolade"]
```
I eksempelet over inneholder variabelen _handleliste_ en liste med 4 elementer.

## Hente ut fra listen

For å hente ut et bestemt element fra en liste kan vi ta utgangspunkt i navnet til variabelen for så å referere til nummeret som elementet har i listen. Når Python skal nummerere så begynner den på 0 og teller oppover.
```python
handleliste = ["Tomater", "Agurk", "Appelsin", "Sjokolade"]
print(handleliste[2]) #printer ut "Appelsin"
```

## Legge til i listen

Om vi ønsker å legge til noe til listen kan vi bruke kommandoen `.append()`.
```python
liste = [] # Lager en liste som er tom
liste.append(5)
liste.append("Hei")
liste.append(2022)
print(liste) # Skriver ut hele listen [5, "Hei", 2022]
```

## Regnemetoder med lister

**Summen av verdiene i en liste:**
```python
liste1 = [1, 2, 3, 4]
summen = sum(liste1)
print(summen) # Skriver ut 10
```

**Telle antall elementer i en liste:**
```python
antall_tall = len(liste1)
print(antall_tall) # Skriver ut 4
```

**Regne ut gjennomsnitt:**
```python
gjennomsnitt = summen/antall_tall
print(gjennomsnitt) # Skriver ut 2.5
```

**Telle antall av en bestemt verdi:**
```python
liste1 = [1, 1, 3, 4]
antall1 = liste1.count(1)
print(antall1) # Skriver ut 2
```

**Sortere verdiene i en liste:**
```python
liste1 = [4, 1, 3, 4]
liste1.sort()
print(liste1) # Skriver ut [1, 1, 3, 4]
```
